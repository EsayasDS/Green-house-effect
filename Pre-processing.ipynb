```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Define the categorical columns to encode (exclude facility_name)
categorical_cols = [
    'city',
    'state',
    'primary_naics_code',
    'industry_type_subparts',
    'industry_type_sectors',
    'unit_type'
]

# 2. Dictionary to store the fitted encoders for later use
encoders = {}

# 3. Loop through columns and encode
for col in categorical_cols:
    le = LabelEncoder()  # Initialize a new encoder for each column
    df[col + '_Encoded'] = le.fit_transform(df[col].astype(str))
    encoders[col] = le  # Store encoder for inverse transform if needed

# 4. Verify result
df.head()
```

```python
df = df.drop(columns=[
    'city', 'state', 'primary_naics_code',
    'industry_type_subparts', 'industry_type_sectors',
    'unit_type'
])
```

```python
# Normalize column names (remove leading/trailing spaces)
df.columns = df.columns.str.strip()

# Create target variable: Total CO2-equivalent
# GWP: CH4 = 25, N2O = 298
df['Total_CO2e'] = (
    df['unit_co2_emissions_non-biogenic']
    + df['unit_biogenic_co2_emissions_metric_tons']
    + df['unit_methane_ch4_emissions'] * 25
    + df['unit_nitrous_oxide_n2o_emissions'] * 298
)
```

```python
from sklearn.preprocessing import MinMaxScaler

# Scale the feature
scaler_capacity = MinMaxScaler()
df['Capacity_Scaled'] = scaler_capacity.fit_transform(
    df[['unit_maximum_rated_heat_input_capacity_mmbtu/hr']]
)

# Scale the target
scaler_target = MinMaxScaler()
df['Total_CO2e_Scaled'] = scaler_target.fit_transform(
    df[['Total_CO2e']]
)
```
