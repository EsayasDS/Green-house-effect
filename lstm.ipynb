```python
col = 'Unit Maximum Rated Heat Input Capacity (mmBTU/hr)'

df[col] = df.groupby('Unit Type')[col].transform(
    lambda x: x.fillna(x.median())
)
```

So here we imputed the missing values by using the median within each Unit Type rather than the global median. This is because heat input capacity is strongly determined by unit type. So we first grouped the data by unit type then calcualted the median Heat Input Capacity for each unit type and filled each missing values with the median of its corresponding Unit Type catagory.

```python
df['Primary NAICS Code'] = df['Primary NAICS Code'].fillna('Unknown')
```

So here I substituted it with "Unknown" because deep learning models use embeddings, and embeddings require every categorical value to be a valid string. This method also identifies unknown categories, so rather than dropping them, imputing is better. You might be wondering why impute an integer with a string — the answer is that even though NAICS codes contain digits, they are labels, not numbers. NAICS codes behave like ZIP codes, phone area codes, or product category IDs; therefore, they are categorical.

```python
# normalize column names to remove accidental leading/trailing whitespace
df.columns = df.columns.str.strip()

# calculate CO2-equivalent (GWP: CH4=25, N2O=298)
df['Total_CO2e'] = (
    df['Unit CO2 emissions (non-biogenic)']
    + df['Unit Biogenic CO2 emissions (metric tons)']
    + df['Unit Methane (CH4) emissions'] * 25
    + df['Unit Nitrous Oxide (N2O) emissions'] * 298
)
```

Here we created this new feature because the green house gases dosent have the same effect. So 1 unit of metahne has the effect of 25 CO2 and one unit of Nitrous Oxide has the effect of 298 CO2. The other thing is it will gving our model one clear target to predict.

```python
from scipy.stats import boxcox

columns_to_transform = [
    'Total_CO2e',
    'Unit Maximum Rated Heat Input Capacity (mmBTU/hr)'
]

transformed_data = {}
lambda_values = {}

# --- 2. Apply Box-Cox Transformation to each column ---
for col in columns_to_transform:
    data_to_transform = df[col]
    
    # Box-Cox requires data to be strictly positive (x > 0). 
    # Check for non-positive values and adjust by adding a small constant if necessary.
    if data_to_transform.min() <= 0:
        # Add a constant (1e-6) plus the absolute minimum to ensure all values are positive
        data_adjusted = data_to_transform + np.abs(data_to_transform.min()) + 1e-6 
    else:
        data_adjusted = data_to_transform

    # Apply Box-Cox transformation and calculate the optimal lambda
    transformed_values, lambda_opt = boxcox(data_adjusted)
    
    # Store results in the DataFrame and dictionary
    new_col_name = f'{col}_BoxCox'
    df[new_col_name] = transformed_values
    lambda_values[col] = lambda_opt

# --- 3. Output Results ---
print("Optimal Lambda Values ($\lambda$):")
for col, lmbda in lambda_values.items():
    print(f"  {col}: {lmbda:.4f}")

print("\nHead of DataFrame with Transformed Columns:")
print(df[[*columns_to_transform, *[f'{col}_BoxCox' for col in columns_to_transform]]].head().to_markdown(index=False))
```

```python

# --- Sequence Generation Function ---
def create_sequences(df, lookback=3): # lookback=3 means using past 3 years to predict the next year
    
    #Transforms the panel data into sequence samples (X, Y) for Deep Learning.
    
    X_dynamic = [] # past 3 years of dynamic features
    X_static = []  # static/categorical features for each series
    Y_target = []  # target variable for the next year or next year emission

    # Features that change over time (dynamic)
    dynamic_features = ['Total_CO2e_Scaled', 'Capacity_Scaled']

    # Features that are constant for a series (static covariates/embeddings)
    static_features = [c for c in df.columns if c.endswith('_Encoded')]

    # Iterate over each unique time series (unit)
    for series_id, group in df.groupby('Series_ID'):
        data = group[dynamic_features].values
        target = group['Total_CO2e_Scaled'].values
        
        # Get the static features (constant for the group)
        static_data = group[static_features].iloc[0].values

        # Apply the sliding window logic
        for i in range(len(data) - lookback):
            # X_dynamic: Past 3 years of dynamic features (e.g., [t-3, t-2, t-1])
            X_dynamic.append(data[i : (i + lookback)])
            
            # Y_target: The target is the next year's emission (t)
            Y_target.append(target[i + lookback])
            
            # X_static: Static data is the same for all time steps in a sequence
            X_static.append(static_data)
            
    return np.array(X_dynamic), np.array(X_static), np.array(Y_target)

# Define the lookback window (T=3 years history to predict T+1)
T = 3
X_dynamic, X_static, Y_target = create_sequences(df, lookback=T)

# Save the arrays
np.savez_compressed('sequence_data.npz', X_dynamic=X_dynamic, X_static=X_static, Y_target=Y_target)
print("Sequences generated and saved!")
print("X_dynamic shape:", X_dynamic.shape)
print("X_static shape:", X_static.shape)
print("Y_target shape:", Y_target.shape)
```

```python
split = int(0.8 * len(X_dynamic))

X_dyn_train = X_dynamic[:split]
X_stat_train = X_static[:split]
Y_train = Y_target[:split]

X_dyn_test = X_dynamic[split:]
X_stat_test = X_static[split:]
Y_test = Y_target[split:]
```

```python
# Dynamic input
print("X_dyn_train shape:", X_dyn_train.shape)
print("X_dyn_test shape:", X_dyn_test.shape)

# Static input
print("X_stat_train shape:", X_stat_train.shape)
print("X_stat_test shape:", X_stat_test.shape)

# Target
print("Y_train shape:", Y_train.shape)
print("Y_test shape:", Y_test.shape)
```

```python
from tensorflow.keras import Input, Model
from tensorflow.keras.layers import LSTM, Dense, concatenate

# Dynamic input
dyn_input = Input(shape=(X_dyn_train.shape[1], X_dyn_train.shape[2]))
x = LSTM(64)(dyn_input)
x = Dense(32, activation='relu')(x)

# Static input
stat_input = Input(shape=(X_stat_train.shape[1],))
s = Dense(32, activation='relu')(stat_input)
s = Dense(16, activation='relu')(s)

# Merge
combined = concatenate([x, s])
z = Dense(32, activation='relu')(combined)
z = Dense(16, activation='relu')(z)
output = Dense(1)(z)

model = Model(inputs=[dyn_input, stat_input], outputs=output)
model.compile(optimizer='adam', loss='mse')

model.summary()
```

```python
history = model.fit(
    [X_dyn_train, X_stat_train],  # Inputs
    Y_train,                      # Targets
    validation_data=([X_dyn_test, X_stat_test], Y_test),  # Validation
    epochs=20,                     # Number of passes over the dataset
    batch_size=64                  # Adjust based on memory
)
```

```python
test_loss = model.evaluate([X_dyn_test, X_stat_test], Y_test)
print("Test Loss (MSE):", test_loss)
```
